```python
from google.colab import files
uploaded = files.upload()
```

```python
import whisper, re, time
model = whisper.load_model("medium")

answers = {
 "01": "아이스 아메리카노 한 잔 주세요",
 "02": "먹고 갈게요",
 "03": "디카페인 돼요",
 "04": "시그니처 메뉴가 뭐예요",
 "05": "와이파이 비밀번호가 뭐예요",
 "06": "화장실 어디에 있어요",
}
def norm(s): return re.sub(r"[^가-힣]", "", s)

for name in sorted(uploaded):
    target = answers.get(name[:2], "")
    t = time.time()
    text = model.transcribe(name, language="ko")["text"].strip()
    sec = time.time() - t
    ok = "맞음" if norm(text) == norm(target) else "틀림"
    print(f"{name} | 정답: {target} | 인식: {text} | {ok} | {sec:.1f}초")
```
